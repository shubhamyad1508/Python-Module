## Operators¶
    Arithmetic Operators : +,-,*,/,power
    Logical Operators : AND,OR,NOT
    Membership Operators : IN, NOT IN
    Conditional Operators : >,<,==,>=,<=
    Assignment Operators : +=,-=,*=,...

```python
# Arithmetic Operators
# Addition

a =10
b= 20
add1 = a + b
print(add1)
```

```python
# Subtraction
sub1 = b-a
print(sub1)
```

```python
# Multiplication
mul1 = a*b
print(mul1)
```

```python
# Division
div1 = a/b
print(div1)
```

```python
# Integer Division
int_div1 = b//a
print(int_div1)
```

```python
# Modulus(Gives reminder)
mod1 = a%b
print(mod1)
```

```python
# power
pow1 = a**2
pow2 = b**3

print(pow1)
print(pow2)
```

```python
location = "Pune"

if location == "pune":
    print("You are in Pune")
else:
    print("You are not in Pune")
```

```python
location3 = input("Enter your location: ")
if (age>=18) and location3 in ("Pune", "Kolahapur", "Mumbai", "Nagpur"):
    print("You are eligible to vote")
else:
 print("not Eligible")
```

```python
location3 = input("Enter your location: ")
if (age>=18) and location3.lower() in ("pune", "kolahapur", "mumbai", "nagpur"):
    print("You are eligible to vote")
else:
 print("not Eligible")
```

```python
location3 = input("Enter your location: ")
if (age>=18) and location3.upper() in ("PUNE", "KOLAHAPUR", "MUMBAI", "NAGPUR"):
    print("You are eligible to vote")
else:
 print("not Eligible")
```

```python
location3 = input("Enter your location: ")
if (age>=18) and location3.title() in ("Pune", "Kolahapur", "Mumbai", "Nagpur"):
    print("You are eligible to vote")
else:
 print("not Eligible")
```
